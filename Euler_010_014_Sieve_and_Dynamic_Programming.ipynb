```python
# Project Euler #10: Summation of Primes
# Algorithm: Sieve of Eratosthenes | Time Complexity: O(n log log n)

def sum_of_primes(limit: int) -> int:
    is_prime = [True] * limit
    is_prime[0] = is_prime[1] = False  # 0 and 1 are not prime

    for p in range(2, int(limit ** 0.5) + 1):
        if is_prime[p]:
            # Mark all multiples of p starting from p^2 as composite
            is_prime[p * p : limit : p] = [False] * len(range(p * p, limit, p))

    return sum(i for i, prime in enumerate(is_prime) if prime)

if __name__ == "__main__":
    LIMIT = 2_000_000
    result = sum_of_primes(LIMIT)
    print(f"Sum of all primes below {LIMIT:,}: {result}")
```

```python
# Project Euler #14: Longest Collatz Sequence
# Algorithm: Dynamic Programming (Memoization) | Time Complexity: O(n) amortized

def longest_collatz(limit: int) -> tuple[int, int]:
    memo = {1: 1}  # Cache known sequence lengths: {number: chain_length}
    max_length = 1
    best_start = 1

    for start in range(2, limit):
        n = start
        steps = 0

        while n not in memo:
            if n % 2 == 0:
                n //= 2
            else:
                n = 3 * n + 1
            steps += 1

        total_length = steps + memo[n]
        memo[start] = total_length

        if total_length > max_length:
            max_length = total_length
            best_start = start

    return best_start, max_length

if __name__ == "__main__":
    LIMIT = 1_000_000
    starting_num, chain_len = longest_collatz(LIMIT)
    print(f"Starting number under {LIMIT:,} with longest chain: {starting_num} (Length: {chain_len})")
```
